## Tools

Models can request to call the perform task such as fetching data from a database, searching the web, or running code. Tools are pairings of:

    1. A schema, includeing the name of the tool, a description, ans/or argument definitions(often a JSON schema)
    2. A function or coroutine to excute.


```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain.chat_models import init_chat_model
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")


model= init_chat_model("groq:llama-3.1-8b-instant")

response=model.invoke("Write a poem about the beauty of nature.")
response
```

```python
# Tools

from langchain.tools import tool

@tool
def get_weather(location:str)->str:
    """Get the current weather of a location"""
    return f"The current weather in {location} is Sunny, 30°C."

model_with_tool= model.bind_tools([get_weather])
```

```python
response=model_with_tool.invoke("What the weather in Bangalore")
print(response)
```

```python
response.tool_calls
```

## Tool Execution Loop

```python
# step1: Model generates tool calls

messages= [{"role":"user","content":"What the weather in Bangalore"}]
ai_msg= model_with_tool.invoke(messages)
messages.append(ai_msg)

# step2: Execute tool calls and collect results

for tool_call in ai_msg.tool_calls:
    #Execute the tool call with generated arguments
    tool_result= get_weather.invoke(tool_call)
    messages.append(tool_result)
    

    
# step3: Model generates final response
final_response= model_with_tool.invoke(messages)
print(final_response.content)
```
